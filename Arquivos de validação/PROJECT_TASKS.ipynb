# PyInvest — Tarefas Prioritárias (Plano de Ação)

Este arquivo detalha tarefas acionáveis para resolver as lacunas do **Modo Expert** e melhorar a robustez do motor Monte Carlo.

## Prioridade: Alta

1. **[MC-001] Expandir `MonteCarloInput`**
   - Arquivo: `core/monte_carlo.py`
   - Descrição: adicionar campos formais: `expert_mode: bool = False`, `simulation_method: str = 'bootstrap'`, `historical_returns: Optional[List[HistoricalReturn]] = None`, `random_seed: Optional[int] = None`.
   - Testes: criar unit test garantindo que os campos existam e aceitem valores.

2. **[MC-002] Integrar métodos Expert no `MonteCarloEngine.run()`**
   - Arquivo: `core/monte_carlo.py`
   - Descrição: quando `inputs.expert_mode == True`, usar `inputs.simulation_method` para gerar matriz de retornos anuais:
     - `bootstrap`: usar `core/statistics.bootstrap_returns` com `[r.return_rate for r in inputs.historical_returns]` (verificar unidade: decimal)
     - `normal`/`t_student`: usar `normal_returns`/`t_student_returns` para gerar matrizes (n_sim x n_years)
   - Converter retornos anuais para taxas mensais, aplicá-las mês-a-mês (vetorizado) para computar `all_balances`.
   - Testes: validar resultados com dados sintéticos e comparar percentis esperados.

3. **[MC-003] Substituir `np.random.seed` por `Generator`**
   - Arquivo: `core/statistics.py` (+ possíveis chamadas em outros módulos)
   - Descrição: usar `rng = np.random.default_rng(seed)` e `rng.choice`/`rng.normal`/`rng.standard_t`.

## Prioridade: Média

4. **[UI-001] Unificar e documentar unidades (%) vs decimal**
   - Arquivo: `ui/window_modern.py` e `core/statistics.py`
   - Descrição: garantir que conversões sejam explícitas e consistentes (por exemplo, `return_rate` sempre armazenado em decimal; UI mostra %).

5. **[TEST-001] Adicionar testes unitários e de integração**
   - Arquivos: `tests/test_monte_carlo.py`, `tests/test_statistics.py`
   - Cobertura: bootstrap, normal, t_student, integração end-to-end com `window_modern` (simulação headless com `MonteCarloInput`).

## Prioridade: Baixa

6. **[DOC-001] Atualizar documentação e tooltips da UI**
   - Arquivo: `README.md`, `ui/*.py` (strings de ajuda)
   - Descrição: explicar requisitos do Bootstrap (mínimo de anos), detalhar implicações de cada método.

## Anotações sobre implementação

- Ao implementar `MC-002`, priorizar implementação vetorizada (evitar loops Python por simulação sempre que possível).
- Considerar limites de memória: `n_simulations * total_months` pode ser grande. Se necessário, apoiar modo 'streaming' (chunks).
